Data Types \
  Data types are a classification of data which tell the compiler how the programmer intends to use the data \
  They determine the operations that can be performed on the data, the values that the data can take, and the amount of memory needed to store the data

```python
# Integer  example 

x = 45
print (type(x))
```

```python
# Float example 

y = 6.34
print(y)
print(type(y))
```

```python
# string example

economy_class = "Rich"
print(economy_class)
print(type(economy_class))
```

```python
# Boolean example

is_rich = True
print(type(is_rich))
print(is_rich)
```

```python
is_rich = bool
print(is_rich)
```

```python
a = 45
b = 45

a==b
```

```python
complex_num = 2 + 3j
print(type(complex_num))
print(complex_num.real)
print(complex_num.imag)
print(complex_num.conjugate())
print(complex_num**2)
print(complex_num)
```

```python
u = 3j
print(type(u))
```

```python
# List: ordered and changeable

numbers_list = [1, 2, 3, 4]
print(type(numbers_list), numbers_list)
```

```python
# Tuple: ordered and immutable
numbers_tuple = (10, 50, 30)
print(type(numbers_tuple), numbers_tuple)
```

```python
# Set: unordered and unique values
numbers_set = {1, 2, 3, 3, 5, 5, 5}
print(type(numbers_set), numbers_set)
```

```python
# Dictionary: key-value pairs
person = {"name": "John Bosko", "age": 25, "is_rich": True}
print(type(person), person)
```

```python
# Range: sequence of numbers
number_range = range(5)
print(type(number_range), list(number_range))
```

```python
# Bytes: stores raw binary data
byte_data = b"hello"
print(type(byte_data), byte_data)
```

```python
# NoneType: represents no value
empty_value = None
print(type(empty_value), empty_value)
```

```python
# Summary of Python Data Types

print("=== Primitive Data Types ===")
print(f"Integer: {x}")
print(f"Float: {y}")
print(f"String: {economy_class}")
print(f"Boolean: {type(True)}")
print(f"Complex: {complex_num}")
print(f"Bytes: {byte_data}")
print(f"NoneType: {empty_value}")

print("\n=== Collection Data Types ===")
print(f"List: {numbers_list}")
print(f"Tuple: {numbers_tuple}")
print(f"Set: {numbers_set}")
print(f"Dictionary: {person}")
print(f"Range: {list(number_range)}")
```
