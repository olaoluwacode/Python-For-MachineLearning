```python
num1 = 34
num2 = 34
print(num1 is num2) # is operator checks if both the variables point to the same object in memory
print(num1>=num2) # greater than or equal to operator
print(num1<=num2) # less than or equal to operator
print(num1!=num2) # not equal to operator
print(num1>num2) # greater than operator
print(num1<num2) # less than operator
print(num1==num2) # equal to operator
print(num1 is not num2) # is not operator checks if both the variables point to different objects in memory
```

```python
# Logical operators with conditional statements
def check_eligibility(age, has_license, has_insurance):
  """Check if someone is eligible to drive"""
  if age >= 18 and has_license and has_insurance:
    return "Eligible to drive"
  elif age >= 18 and (has_license or has_insurance):
    return "Partially eligible - missing some requirements"
  else:
    return "Not eligible to drive"

# Test the function
print(check_eligibility(25, True, True))
print(check_eligibility(20, True, False))
print(check_eligibility(16, True, True))

# Using not operator
def is_invalid(value):
  """Check if a value is invalid"""
  return not value

print(is_invalid(True))
print(is_invalid(False))
```

```python
x = True
y = True
print(x and y) # and operator returns True if both the operands are True
print(x or y) # or operator returns True if at least one of the operands is True
print(not x) # not operator returns True if the operand is False
print(not y) # not operator returns True if the operand is False
```
